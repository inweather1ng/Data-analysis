```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# 1. FIX FILE PATHS (Using underscores as seen in your previous prompts)
# We use skiprows=2 for rainfall and skiprows=0 for folsom (adjust if needed)
try:
    precip_data = pd.read_csv("../Raw_data/placer county rainfall.csv", skiprows=2)
    folsom = pd.read_csv("../Raw_data/reservoir data.csv")
    print("Files loaded successfully!")
except FileNotFoundError:
    print("Error: Check your folder names! Is it 'Raw_data' or 'raw+data'?")

# 2. FORCE NUMERIC (The Y-Axis Fix)
# We use .iloc[:, 1] to grab the 2nd column regardless of its name
folsom.iloc[:, 1] = pd.to_numeric(folsom.iloc[:, 1].astype(str).str.replace(',', ''), errors='coerce')
precip_data.iloc[:, 1] = pd.to_numeric(precip_data.iloc[:, 1].astype(str).str.replace(',', ''), errors='coerce')

# 3. DATE CONVERSION (The Timeline Fix)
# We tell pandas exactly what the column index is
folsom.iloc[:, 0] = pd.to_datetime(folsom.iloc[:, 0], errors='coerce')
precip_data.iloc[:, 0] = pd.to_datetime(precip_data.iloc[:, 0], errors='coerce')

# 4. CLEAN UP
# Remove any rows that became empty (NaN) during conversion
folsom = folsom.dropna().sort_values(folsom.columns[0])
precip_data = precip_data.dropna().sort_values(precip_data.columns[0])

# 5. THE GRAPH
fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.08,
                    subplot_titles=("Precipitation (Inches)", "Reservoir Storage (Acre-Feet)"))

# Rainfall Trace
fig.add_trace(
    go.Bar(x=precip_data.iloc[:, 0], y=precip_data.iloc[:, 1], name="Rain"),
    row=1, col=1
)

# Reservoir Trace
fig.add_trace(
    go.Scatter(x=folsom.iloc[:, 0], y=folsom.iloc[:, 1], name="Storage", fill='tozeroy'),
    row=2, col=1
)

fig.update_layout(height=800, title_text="Folsom Watershed Analysis", showlegend=False)
fig.show()

```
